```python
n = int(input("Enter the n: "))
k = int(input("Enter the k: "))
j = int(input("Enter the j: "))
m = int(input("Enter the m: "))
p = int(input("Enter the p: "))

if n <= 0 or k <= 0 or j <= 0 or m < 0 or p < 0:
    print("INVALID INPUT")
else:
    x = (m // k) + (p // j)

    if m % k != 0 or p % j != 0:
        x = x + 1

    if x > n:
        x = n

    n = n - x

    print("Number of Monkeys left on the tree:", n)
```
