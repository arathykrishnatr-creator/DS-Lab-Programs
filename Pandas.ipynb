```python
import matplotlib.pyplot as plt
year = [2001, 2002, 2003, 2004, 2005, 2006, 2007]
value = [24000, 22500, 19700, 14500, 10000, 5800, 3000]
plt.plot(year, value,color="red",marker="*",linestyle="--",linewidth=2,
         markersize=10,markerfacecolor="green",markeredgecolor="green")
plt.title("Value of Car")
plt.xlabel("Year")
plt.ylabel("Value")
plt.grid(True)

plt.show()
```

```python
import matplotlib.pyplot as plt
months = ["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct",
          "Nov","Dec"]
affordable = [173,153,195,147,120,144,148,109,174,130,172,131]
luxury = [189,189,105,112,173,109,151,197,174,145,177,161]
super_luxury = [185,185,126,134,196,153,112,133,200,145,167,110]
plt.scatter(months, affordable, color="pink", label="Affordable")
plt.scatter(months, luxury, color="yellow", label="Luxury")
plt.scatter(months, super_luxury, color="blue", label="Super Luxury")
plt.title("Sales Data")
plt.xlabel("Months of Year", fontsize=18)
plt.ylabel("Sales of Segments")
plt.legend()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("/iris.csv")
plt.scatter(df["sepal_length"], df["sepal_width"])
plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("/iris.csv")
colors = {"setosa":"red","versicolor":"green","virginica":"blue"}
for flower in colors:
    data = df[df["species"] == flower]
    plt.scatter(data["sepal_length"], data["sepal_width"],
                color=colors[flower], label=flower)
plt.legend()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("/iris.csv")
plt.subplot(2,2,1)
plt.hist(df["sepal_length"])
plt.subplot(2,2,2)
plt.hist(df["sepal_width"])
plt.subplot(2,2,3)
plt.hist(df["petal_length"])
plt.subplot(2,2,4)
plt.hist(df["petal_width"])
plt.tight_layout()
plt.show()
```
